```python
from rdkit import Chem
from rdkit.Chem import AllChem
```

```python
diene=Chem.MolFromSmiles("C=CC=C")
dienophile = Chem.MolFromSmiles("O=C1OC(=O)C=C1")
if diene is None or dienophile is None:
    print("Error: Invalid SMILES input!")
else:
    print("Diene heavy atoms:", diene.GetNumAtoms())
    print("Dienophile heavy atoms:", dienophile.GetNumAtoms())
```

```python
rxn=AllChem.ReactionFromSmarts("[C:1]=[C:2]-[C:3]=[C:4].[C:5]=[C:6]>>[C:1]1-[C:2]=[C:3]-[C:4]-[C:5]-[C:6]1")
products = rxn.RunReactants((diene, dienophile))
print (len(products))
print(Chem.MolToSmiles(products[0][0]))
```

```python
unique_products = set()
for prod_tuple in products:
    mol = prod_tuple[0]
    Chem.SanitizeMol(mol)
    unique_products.add(Chem.MolToSmiles(mol))
print(unique_products)
```

```python
# Two AI-generated product predictions to evaluate
ai_guess_1 = "C1=CCC2C(=O)OC(=O)C2C1"  # Valid product written in non-canonical SMILES order
ai_guess_2 = "O=C1OC(=O)C2CC=CC12"     # Hallucinated 5-membered ring (wrong cycloaddition)

# Check Prediction 1
mol_1 = Chem.MolFromSmiles(ai_guess_1)
if mol_1 is not None:
    canon_1 = Chem.MolToSmiles(mol_1)
    print("Prediction 1 valid?", canon_1 in unique_products)
else:
    print("Prediction 1 failed SMILES parsing.")

# Check Prediction 2
mol_2 = Chem.MolFromSmiles(ai_guess_2)
if mol_2 is not None:
    canon_2 = Chem.MolToSmiles(mol_2)
    print("Prediction 2 valid?", canon_2 in unique_products)
else:
    print("Prediction 2 failed SMILES parsing.")
```
